```python
import pandas as pd
filepathh = '/content/drive/MyDrive/Colab Notebooks/_data.csv'

def run_data_pipeline(input_path:str,output_path:str,summary_path:str):
 df  = pd.read_csv(input_path)
 numeric_col =df.select_dtypes(include=["number"]).columns
 text_cols = df.select_dtypes(include=["object","category"]).columns
 if len(numeric_col)>0:
   df[numeric_col]= df[numeric_col].fillna(df[numeric_col].median())
 if len(text_cols) >0:
   df[text_cols] = df[text_cols].fillna("unknown")

 df = df.dropna(how ="all")
 df.to_csv(output_path ,index = False)

 group_cols =text_cols[0] if len(text_cols)>0 else df.columns[0]

 firstnum_col = (numeric_col[0]
                 if len(numeric_col)>0 else (df.columns[1] if len(df.columns)> 1 else df.columns[0]))

 sum_df = (df.groupby(group_cols)[firstnum_col].agg(totalcount='count'
 ,  mean_value ='mean',
   max_val='max'
  ,min_val='min').reset_index())
 sum_df.to_csv(summary_path,index=False)
 print(f"Pipeline complete!\n - Cleaned data -> {output_path}\n - Summary report -> {summary_path}")

 run_data_pipeline('input_data.csv', 'cleaned_data.csv', 'summary_report.csv')
```
